```python
superstore = pd.concat([df_3, df_4], ignore_index=True)
print(len(superstore))
```

```python
customers = superstore[['Customer ID', 'Customer Name', 'Segment']]. drop_duplicates()
orders = superstore[['Row ID', 'Order ID', 'Customer ID', 'Sales', 'Profit']]

left_merge = orders.merge(customers, on= 'Customer ID', how= 'left')
print(len(orders), len(left_merge))
```

```python
left_merge.to_sql('left_merge', con= engine, if_exists = 'replace', index = False)
```

```python
right_merge = orders.merge(customers, on= 'Customer ID', how= 'right')
print(len(orders), len(right_merge))
```

```python
right_merge.to_sql('right_merge', con= engine, if_exists = 'replace', index = False)
```

```python
inner_merge = orders.merge(customers, on= 'Customer ID', how= 'inner')
print(len(orders), len(inner_merge))
```

```python
inner_merge.to_sql('inner_merge', con= engine, if_exists = 'replace', index = False)
```

```python
check_df_1 = pd.read_sql('select * from left_merge', con = engine)
print(check_df_1);
```

```python
print("Orders:", len(orders))
print("Left Merge:", len(left_merge))
print("Inner Merge:", len(inner_merge))
print("Right Merge:", len(right_merge))
```
